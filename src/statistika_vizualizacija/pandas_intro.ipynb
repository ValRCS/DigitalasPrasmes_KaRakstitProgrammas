```python
# pārbaudīsim Pandas versiju
import pandas as pd
print("Pandas versija:", pd.__version__)
# ja šī komanda izvada versiju, tad Pandas ir veiksmīgi instalēts un gatavs lietošanai!
# otrs moments, tagad pd ļaus mums izmantot Pandas funkcijas, piemēram, pd.read_csv() datu ielādei no CSV faila.
```

```python
# jā pandas nav tad uzstādam ar sekojošo komandu:
# pip install pandas
# mums vajag arī lasīt un rakstīt Excel failus, tāpēc uzstādam arī openpyxl bibliotēku:
# pip install openpyxl
# rakstīšanā Excel failā:
# xlslwriter arī
# pip install xlsxwriter
# kopā vienā rindā:
# pip install pandas openpyxl xlsxwriter

# nolasīsism terini.csv failu un izvadīsim to uz ekrāna
# svarīgi lai terini.csv būtu tajā pašā mapē kur atrodas šis Python fails, pretējā gadījumā būs jānorāda pilns ceļš līdz failam
df = pd.read_csv("terini.csv") # nolasām csv failu
# df ir ļoti populārs saīsinājums datu rāmim (dataframe)
print(df) # izvadām datu rāmi uz ekrāna
```

```python
# var būt ērtāk izvadīt tikai galvu - head
# otrs print nav obligāts Jupyter Notebook šūnas pēdejā rindā, jo Jupyter automātiski izvada pēdējo objektu, bet tas ir labs ieradums, lai kods būtu saprotamāks un strādātu arī ārpus Jupyter vidē.
df.head() # izvadām datu rāmi uz ekrāna, bet tikai pirmās 5 rindas
```

```python
# aste
# pēdejās 3 rindas
df.tail(3)
```

```python
# describe mums dod statistiku par skaitliskajiem datiem, piemēram, vidējo vērtību, minimumu, maksimumu, standartnovirzi utt.
df.describe()
```

```python
# šo analīze mēs varam uzreiz arī saglabāt dažādos formātos, piemēram, Excel failā vai CSV failā
# pagaidām tikai CSV
df.describe().to_csv("terini_statistika.csv", index=True) # saglabājam statistiku CSV failā, index=True nozīmē, ka saglabāsim arī indeksu (kolonnas nosaukumus)
```

```python
# līdzigi var saglabāt arī Excel failā, bet tam ir jānorāda arī engine, jo pēc noklusējuma Pandas izmanto openpyxl, bet mums ir uzstādīts arī xlsxwriter, tāpēc jānorāda kurš engines izmantot:
df.describe().to_excel("terini_statistika.xlsx", index=True, engine="xlsxwriter") # saglabājam statistiku Excel failā, index=True nozīmē, ka saglabāsim arī indeksu (kolonnas nosaukumus), engine="xlsxwriter" nozīmē, ka izmantosim xlsxwriter bibliotēku Excel faila rakstīšanai
```
