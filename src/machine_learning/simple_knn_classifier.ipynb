# Vienkāršs Klasifikators ar Scikit-learn

## Kas ir scikit-learn?

Scikit-learn ir populāra Python bibliotēka, kas nodrošina vienkāršus un efektīvus rīkus datu analīzei un mašīnmācībai. Tā piedāvā plašu klāstu algoritmu gan klasifikācijai, gan regresijai, kā arī dažādus rīkus datu apstrādei un modelēšanai.

Mājas lapa: [https://scikit-learn.org/](https://scikit-learn.org/)

```python
# importēsim KNeighbors klasifikatoru no scikit-learn bibliotēkas
from sklearn.neighbors import KNeighborsClassifier
# importējam konkrēto KNeighborsClassifier klasi no scikit-learn bibliotēkas
# X šeit varētu simboližet datu pārus
# kur pirmais vecums un otrs stundas dienā pie datora :)
X = [[10,2],[12,3],[14,5],[16,6]]
# mums vajag arī atbildes
y = ["spēles","spēles","programmēšana","programmēšana"]
# piezīme īsta mašīnmācīsānas projektā mums vajadzētu stipri vairāk datu, 
# bet šeit mēs tikai demonstrējam kā strādā KNeighborsClassifier

# izveidosim KNeighborsClassifier objektu
model = KNeighborsClassifier(n_neighbors=3)
# n_neighbors ir parametri, kas nosaka, cik tuvāko kaimiņu nepieciešams apskatīt, lai noteiktu klasi

# mums ir modelis bet tas nav trenēts, tāpēc mēs to trenēsim ar mūsu datiem
model.fit(X, y) # vienā rindiņā notiek liels darbs - model.fit() metode trenē modeli, izmantojot mūsu datu pārus X un atbildes y

# varam jau tagad sākt lietot modeli, lai prognozētu klasi jauniem datiem
# piemēram ja ir 13 gadus vecs un pavada 4 stundas dienā pie datora, kāda būs viņa aktivitāte?
new_data = [[13,4]]
prediction = model.predict(new_data) # model.predict() metode izmanto trenēto modeli, lai prognozētu klasi jauniem datiem

print("Ieteiktā aktivitāte:", prediction[0]) # izdrukājam prognozēto klasi, šajā gadījumā tas būs "spēles" vai "programmēšana" atkarībā no datiem un KNeighborsClassifier darbības
```
